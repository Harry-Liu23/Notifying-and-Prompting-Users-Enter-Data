```python
from scipy.stats import shapiro
from scipy.stats import ks_2samp

w1_intel = [25,13,25,10,11,9,59,55,20,13,21,45,7,19,13,8,24,40,5,6,17,5,5]

stat, p = shapiro(w1_intel)

print('Shapiro-Wilk Test statistic:', stat)
print('p-value:', p)
```

```python
w1_trad = [87, 259, 43, 51, 102, 31, 31]

stat, p = shapiro(w1_trad)

print('Shapiro-Wilk Test statistic:', stat)
print('p-value:', p)
```

```python
from scipy.stats import mannwhitneyu

w1_intel = [25,13,25,10,11,9,59,55,20,13,21,45,7,19,13,8,24,40,5,6,17,5,5]
w1_trad = [87, 259, 43, 51, 102, 31, 31]

stat, p = mannwhitneyu(w1_intel, w1_trad, alternative='two-sided')

print('Mann-Whitney U Test statistic:', stat)
print('p-value:', p)
```
